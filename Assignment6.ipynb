```python
#Question 1 Part A and B
from scipy.stats import norm
#define distribution parameters
mean = 750
std_dev = 60
prob_between= norm.cdf(690, loc=mean, scale=std_dev)-norm.cdf(810, loc=mean,scale=std_dev)
print(f"Probability between 690 and 810: {prob_between:.4f}")
prob_greater_than_180=1- norm.cdf(180, loc=mean, scale=std_dev)
print(f"Probability greater than 180: {prob_greater_than_180:.4f}")
#Question 2
from scipy.stats import norm
#(mean=750, std=29) at the peak(x=750)
density = norm.pdf(750, loc=750, scale=29)
print(f"Density at x=750: {density:.4f}")
```
